## 2. Visualize with UMAP

We cannot plot 768 dimensions at once. **UMAP** squeezes all the dimensions into 2 new columns, `UMAP1` and `UMAP2`. Similar images wil have similar `UMAP1` and `UMAP2` values.

Run the cell. It makes the 2 UMAP columns. (It takes a few seconds.)

````r
set.seed(1)

umap_coords <- recipe(label ~ ., data = select(malaria, -id)) |>
    step_umap(all_predictors(), num_comp = 2) |>
    prep() |>
    bake(new_data = NULL)

umap_coords |> head()
````

**Your turn:** plot `UMAP1` against `UMAP2`. Color the points by `label`.

````r
# Chart code goes here
options(repr.plot.res = 200, repr.plot.height = 5, repr.plot.width = 8)

ggplot(umap_coords, aes(x = UMAP1, y = UMAP2, color = label)) +
  geom_point(alpha = 0.65, size = 1.5) +
  scale_color_manual(
    values = c(
      "uninfected" = "#2C7FB8",
      "parasitized" = "#E34A33"
    ),
    labels = c("Uninfected", "Parasitized")
  ) +
  labs(
    title = "UMAP Visualization of Malaria Blood Cell Embeddings",
    subtitle = "2,000 blood cell images encoded using DINOv2",
    x = "UMAP Dimension 1",
    y = "UMAP Dimension 2",
    color = "Cell Type"
  ) +
  theme_minimal(base_size = 13) +
  theme(
    plot.title = element_text(face = "bold"),
    legend.position = "right"
  )
````

**Check:** do the two kinds of cell land in different places?

- Do you see clear groups? Are they perfectly separate?
- Which cells are hard to tell apart?

The UMAP visualization shows two distinct groups of blood cells based on their embeddings. Most uninfected cells cluster on the left, while most parasitized cells cluster on the right, suggesting that the pretrained DINOv2 encoder captures meaningful differences between the two cell types. However, the groups are not perfectly separated, as some uninfected cells appear within the parasitized cluster and vice versa. Cells in these overlapping regions are harder to distinguish because their embeddings are similar despite having different infection labels.

## 3. Split the data

Run this cell. Do not change it.

- `train`: half of the images. The model learns from these.
- `test`: the other half. The model has never seen these. We measure accuracy here.

````r
set.seed(512)

split <- initial_split(malaria, prop = 0.5, strata = label)
train <- training(split) |> select(-id)
test  <- testing(split)  |> select(-id)

nrow(train)
````

## 4. Pick a model

Pick a classifier model that we've worked with before (e.g. random forest). Fit it on `train`. Find the accuracy on `test`.

Options:

- `logistic_reg(penalty = 0.01, mixture = 0) |> set_engine('glmnet')`: logistic regression. We have 768 columns, so it needs a `penalty` (like in lecture 16).
- `rand_forest(trees = 200) |> set_mode('classification') |> set_engine('ranger')`: random forest.
- `mlp(hidden_units = 10, epochs = 500) |> set_mode('classification') |> set_engine('nnet', MaxNWts = 10000)`: neural net.

Use the formula `label ~ .` (all other embeddings columns are predictors).

````r
# Fit your model here

set.seed(512)

fit_rf <- rand_forest(trees = 200) |>
  set_mode("classification") |>
  set_engine("ranger") |>
  fit(label ~ ., data = train)

fit_rf
````

<details>
<summary>Hint: how do I find the accuracy?</summary>

1. `predict()` gives the predicted class for each test image.
2. `bind_cols()` puts it next to the true `label`.
3. `accuracy()` compares them.

```r
predict(fit, new_data = test) |>
    bind_cols(test) |>
    accuracy(truth = label, estimate = .pred_class)
```

Here `fit` is the name of your fitted model. Use your own name.

</details>

````r
# Accuracy on the test set
# Accuracy on the test set
test_results <- predict(fit_rf, new_data = test) |>
  bind_cols(test)

test_accuracy <- test_results |>
  accuracy(truth = label, estimate = .pred_class)

test_accuracy
````

````r
# Accuracy on the training set
train_results <- predict(fit_rf, new_data = train) |>
  bind_cols(train)

train_accuracy <- train_results |>
  accuracy(truth = label, estimate = .pred_class)

train_accuracy
````

**Check:** a good model gets about 0.93 to 0.95 on the test images.

- Why did you pick this model?
- Is the accuracy on `train` higher than on `test`? What does that tell you?

I chose a random forest with 200 trees because it can capture complex, nonlinear relationships among the 768 embedding features. The model achieved 100% training accuracy and 93.8% test accuracy. The 6.2 percentage-point difference suggests some overfitting, since the model performs better on the images it was trained on. However, the high test accuracy indicates that the model generalizes well to unseen images and that the DINOv2 embeddings contain useful information for distinguishing parasitized from uninfected blood cells.

## 5. Bonus: raw pixels

We also have the raw pixels for the same cells: `malaria_pixels.csv.gz`. That is 32 x 32 = 1,024 columns named `p1` to `p1024`.

Use the same split. Fit the same kind of model on the pixels. How does the test accuracy compare to the embeddings?

````r
pixels <- readr::read_csv('https://raw.githubusercontent.com/UNC-BIOS-512/jupyterlite/main/data/malaria_pixels.csv.gz', show_col_types = FALSE) |>
    mutate(label = factor(label, levels = c('uninfected', 'parasitized')))

# Use the same split: train_pix has the same images as train
train_pix <- pixels |> filter(id %in% training(split)$id) |> select(-id)
test_pix  <- pixels |> filter(id %in% testing(split)$id)  |> select(-id)



# Fit and evaluate here

# Fit random forest on raw pixels
set.seed(512)

fit_rf_pixels <- rand_forest(trees = 200) |>
  set_mode("classification") |>
  set_engine("ranger") |>
  fit(label ~ ., data = train_pix)

# Test accuracy
pixel_test_accuracy <- predict(fit_rf_pixels, new_data = test_pix) |>
  bind_cols(test_pix) |>
  accuracy(truth = label, estimate = .pred_class)

pixel_test_accuracy
````

*Your answer:* Which is better for finding malaria: pixels or embeddings? Why?

DINOv2 embeddings are better than raw pixels for identifying malaria-infected blood cells. Using the same random forest model, the embeddings achieved 93.8% test accuracy, compared to only 72.2% using raw pixels, a difference of 21.6 percentage points. This suggests that the pretrained encoder captures meaningful visual features that help distinguish infected from uninfected cells. Raw pixels contain basic image information but do not directly represent these higher-level patterns, making classification more difficult.
